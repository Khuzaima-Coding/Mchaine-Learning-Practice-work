## Our Model

```python
def line(x,m=0,b=0):
    y=m*x+b    #y=mx + c  #y=7.7x - 347.5
    return y
```

## Cost Function:

```python
#Lets we finding the error difference between predicted and actual
def mean_squared_error(y_true,y_pred):
    s=(y_true - y_pred)**2
    return s.mean()
```

```python
X=df[["Height"]].values
y_true=df[["Weight"]].values
```

```python
y_pred = line(X,7.7,-347.5) #y=mx + b
df.plot(kind="scatter",x="Height",y="Weight",title="Weight and Height in Adults")

plt.plot([55,78],[75,250],color='yellow',linewidth=3)

plt.plot(X,y_pred, color="red",linewidth=3)
plt.show()

#The line is drawn to be likely best fitted line
```

```python
# y=7.7x-347.5
```

```python
mean_squared_error(y_true,y_pred)**.5  #we muliply it by 0.5 for finding the actual value(by taking root) ,because this is squared error
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(X,y_pred)
```

```python
lr.coef_
```

```python
lr.intercept_
```
