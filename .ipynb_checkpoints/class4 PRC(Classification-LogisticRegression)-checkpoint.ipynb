```python
import pandas as pd
```

```python
data=pd.read_csv("diabetes.csv")
```

```python
data.isna().sum()
```

```python
data.info()
```

```python
data.shape
```

```python
data.head()
```

```python
data.Outcome.unique()   # Two categories means binary classification
```

```python
X=data.drop("Outcome",axis=1)
y=data.Outcome
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=123)
```

```python
from sklearn.linear_model import LogisticRegression
log_reg = LogisticRegression()
```

```python
log_reg.fit(X_train,y_train)
```

```python
y_pred_train=log_reg.predict(X_train)
```

```python
y_pred_train
```

```python
count=0
for o,p in zip(y_train,y_pred_train):
    if o==p:
        count+=1
accuracy=f"Accuracy:{(count/len(y_train)) *100}%"
accuracy
```

```python
list(zip(y_train,y_pred_train))
```

```python
y_train.shape
```

```python
y_pred_test=log_reg.predict(X_test)
```

```python
y_pred_test
```

```python
count=0
for o,p in zip(y_test,y_pred_test):
    if o==p:
        count+=1
accuracy=f"Accuracy:{(count/len(y_test)) *100}%"
accuracy
```

```python
data.Outcome.value_counts()
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_train,y_pred_train)
```

```python
accuracy_score(y_test,y_pred_test)
```

```python
df=pd.read_csv("diabetes.csv")
```

```python
df.corr()
```

```python
X=df.drop(["BloodPressure","SkinThickness","Outcome"],axis=1)
y=df.Outcome
```

```python
from imblearn.over_sampling import SMOTE
smote = SMOTE(sampling_strategy='minority', random_state=42)
X,y = smote.fit_resample(X,y)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.30,random_state=123)
```

```python
#Lets first we can scale this data
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler=MinMaxScaler()
```

```python
scaler.fit(X_train)
```

```python
X_scaled_train=scaler.transform(X_train)
X_scaled_test=scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logi_reg=LogisticRegression()
```

```python
logi_reg.fit(X_scaled_train,y_train)
```

```python
y_pred_train=logi_reg.predict(X_scaled_train)
```

```python
y_pred_test=logi_reg.predict(X_scaled_test)
```

```python
from sklearn.metrics import classification_report


print(classification_report(y_train,y_pred_train))
```

```python
print(classification_report(y_test,y_pred_test))
```
