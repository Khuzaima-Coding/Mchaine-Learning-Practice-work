```python
data.drop_duplicates(inplace=True)
```

```python
data.shape
```

```python
x= data.drop(["target"], axis = 1)
y= data.target
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train,y_test = train_test_split(x,y, test_size=0.15, random_state=42)
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
minmax=MinMaxScaler()     #   0,1
```

```python
min_x_train = minmax.fit_transform(X_train)
min_x_test = minmax.transform(X_test)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
import warnings
warnings.filterwarnings("ignore")
```

### Logictics regressuon

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
lr = LogisticRegression()
lr.fit(min_x_train, y_train)
y_pred = lr.predict(min_x_train)
```

```python
print(classification_report(y_train, y_pred))
```

### Decison Tree

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dr = DecisionTreeClassifier(criterion="entropy")
```

```python
dr.fit(min_x_train, y_train)
```

```python
y_train_pred = dr.predict(min_x_train)
```

```python
print(classification_report(y_train, y_train_pred))
```

```python
y_test_pred_dr = dr.predict(min_x_test)
```

```python
print(classification_report(y_test, y_test_pred_dr))
```
