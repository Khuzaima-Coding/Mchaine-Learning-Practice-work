```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder()
encoded = ohe.fit_transform(data.select_dtypes('O').drop('churn',axis=1))
```

```python
encoded_categories = pd.DataFrame(data = encoded.toarray(),columns=ohe.get_feature_names_out())
```

```python
encoded_categories
```

```python
numbered_data = data.select_dtypes('number')
numbered_data
```

```python
df = pd.concat([numbered_data,encoded_categories],axis=1)
```

```python
df = df.astype(float)
```

```python
df
```

```python
X=df
y = data.churn.replace(['Yes', 'No'],[1,0])
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
lr  = LogisticRegression()
lr.fit(X_train,y_train)
```

```python
pred_y = lr.predict(X_train)
```

```python
from sklearn.metrics import confusion_matrix, classification_report
```

```python
confusion_matrix(y_train,pred_y)
```

```python
y_train.value_counts()
```

```python
pred_y_test=lr.predict(X_test)
```

```python
confusion_matrix(y_test,pred_y_test)
```
