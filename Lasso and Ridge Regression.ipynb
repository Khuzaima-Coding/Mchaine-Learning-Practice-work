```python
import pandas as pd
from sklearn.datasets import fetch_california_housing
```

```python
california_housing=fetch_california_housing(as_frame=True) #as_frame =True can converts our data into DataFrame
california_housing
```

```python
X=california_housing.data
X.sample(20)
```

```python
y=california_housing.target
y
```

```python
california_housing.feature_names
```

```python
print(california_housing.DESCR)  #DECR is the attribute that prints the description of the dataset and connects wth the function(fetch_california_housing),that helps us for modeling.
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.30,random_state=0)
```

### Additional Step: Scaling the Data:Standard Scaler

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)            #we applying this because we see in the data that we have differnent ranges there in our data.
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

lr=LinearRegression()
lr.fit(X_train,y_train)
y_train_pred=lr.predict(X_train)
y_test_pred=lr.predict(X_test)
print(mean_absolute_error(y_train,y_train_pred))
lr.score(X_test,y_test)  #Score is the builtin function of our model
```

```python
#Lets we can applying Ridge regression there
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
#Model
ridge=Ridge()
#Model Train
ridge.fit(X_train,y_train)
#Model Prediction
y_train_pred=ridge.predict(X_train)
#Error
print(mean_absolute_error(y_train,y_train_pred))
print(ridge.score(X_test,y_test))
```
