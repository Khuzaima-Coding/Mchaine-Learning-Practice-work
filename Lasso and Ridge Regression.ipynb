## Grid Search

```python
from sklearn.model_selection import GridSearchCV
#List of Hyperparameters
param_grid={"alpha":[0.0001,0.001,0.01,0.1,1,10,100,1000]}
# model ridge with grid search
ridge_cv=GridSearchCV(ridge,param_grid=param_grid,cv=2)  #cv is cross validation 
#Model train
ridge_cv.fit(X_train,y_train)
#model predict on training data 
y_pred_train=ridge_cv.predict(X_train)
#model error on training data
print(mean_absolute_error(y_train,y_pred_train))
print(mean_squared_error(y_train,y_pred_train))
#model performance on train data
print(r2_score(y_train,y_pred_train))
#model best hyperparameter on training data
print(ridge_cv.best_estimator_)
```
