```python
import joblib
```

```python
model=joblib.load("slr.joblib")
```

```python
model.predict([[73]])  #inches can returns weight in pounds
```

```python
def predict_weight(height):
    return model.predict(height)*0.45   # We can multiply it by 0.45 because 1 pound is equals to 0.45 Kg
```

```python
height_inch=float(input("Enter Your Height in foot:"))*12    #We mulpiply it by 12 for converting foot input into inches ,for prediction

predict_weight([[height_inch]])  # We give data in two dimensional array form 
```
