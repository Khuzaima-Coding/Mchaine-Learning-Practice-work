```python
import pandas as pd
```

```python
house=pd.read_excel("Bengaluru_House_Data.xlsx")
```

```python
house.head()
```

```python
len(house.area_type.unique())
```

```python
len(house.availability.unique())
```

```python
house.info()
```

```python
len(house.location.unique())
```

```python
len(house.society.unique())
```

```python
house.drop(["availability","location","society"],axis=1,inplace=True)
```

```python
house.head()
```

```python
house.isna().sum()
```

```python
house.fillna({"size":house['size'].mode()[0],
             "bath":house.bath.median(),
             "balcony":0},inplace=True)
```

```python
house["BHK"]=house["size"].apply(lambda x:x.split(" ")[0]).astype(int)
```

```python
house.head()
```

```python
house.drop("size",axis=1,inplace=True)
```

```python
house.total_sqft.unique()
```

```python
house=house[house.total_sqft.astype(str).str.isnumeric()]
```

```python
house.total_sqft=house.total_sqft.astype(float)
```

```python
from sklearn.preprocessing import OneHotEncoder,LabelEncoder
```

```python
ohe=OneHotEncoder()
```

```python
area_type_encoded=ohe.fit_transform(house[["area_type"]]).toarray()
```

```python
area_type_encoded=pd.DataFrame(area_type_encoded,columns=ohe.get_feature_names_out())
```

```python
house=pd.concat([house,area_type_encoded],axis=1)
```

```python
house.drop("area_type",axis=1,inplace=True)
```

```python
house.head()
```

```python
house.corr()
```

### Split Data

```python
house.dropna(inplace=True)
```

```python
house.columns
```

```python
X=house.drop(["price"],axis=1)
y=house.price
```

```python
X.shape
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=93)
```

### Feature Scaling 

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc=StandardScaler()
```

```python
sc.fit(X_train)
```

```python
X_train=sc.transform(X_train)
X_test=sc.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score,mean_absolute_error
```

```python
#training 
model =LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
y_pred_train=model.predict(X_train)
```

```python
r2_score(y_train,y_pred_train)
```

```python
mean_absolute_error(y_train,y_pred_train)
```

```python
y_pred_test=model.predict(X_test)
```

```python
r2_score(y_test,y_pred_test)
```

```python
mean_absolute_error(y_test,y_pred_test)
```
